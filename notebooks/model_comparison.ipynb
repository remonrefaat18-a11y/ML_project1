```python
import pandas as pd

dt=pd.read_csv("../results/decisiontree_results.csv")
knn=pd.read_csv("../results/KNN_results.csv")
rf=pd.read_csv("../results/randomforest_results.csv")
nb=pd.read_csv("../results/naivebayes_results.csv")
lr=pd.read_csv("../results/logisticregression_results.csv")
svm=pd.read_csv("../results/svm_results.csv")


comparison = pd.concat(
    [dt, knn, rf,nb, lr, svm],
    ignore_index=True
)

comparison.sort_values(
    by="Accuracy",
    ascending=False
).set_index("Model").style.background_gradient(cmap="Greens").set_caption("Comparison of different models")
```

```python
import matplotlib.pyplot as plt

comparison.plot(
    x="Model",
    y=["Accuracy", "Precision", "Recall", "F1 Score", "ROC AUC"],
    kind="bar",
    figsize=(12,6)
)

plt.title("Models Comparison")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.show()
```

```python
ax = comparison.plot(
    x="Model",
    y="Accuracy",
    kind="bar",
    figsize=(8,6)
)

plt.ylim(0.5,1)

for container in ax.containers:
    ax.bar_label(container, fmt="%.4f")

plt.show()
```

```python
ax2 = comparison.plot(
    x="Model",
    y="ROC AUC",
    kind="bar",
    figsize=(8,6)
)

plt.ylim(0.5,1)

for container in ax.containers:
    ax.bar_label(container, fmt="%.4f")

plt.show()
```
